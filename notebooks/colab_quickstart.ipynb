# Fashion Vision Pipeline: Google Colab Runner (Docker-Free)

This notebook demonstrates how to run the event-driven Fashion Vision Pipeline entirely within a Google Colab virtual machine without Docker.

### Pipeline Components
- **PostgreSQL 16**: Job state persistence and `LISTEN`/`NOTIFY` events.
- **MinIO Object Storage**: S3-compatible claim-check storage for raw images and piece crops.
- **Apache Kafka (KRaft)**: Event-driven broker for Stage 1 and Stage 2 image jobs without Zookeeper.
- **FastAPI Gateway**: Asynchronous ingestion endpoint (`/api/v1/jobs` and `/healthz`).
- **Fashion GPU Worker**: Background daemon running SegFormer and FashionCLIP inference.
